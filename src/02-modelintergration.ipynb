## Streaming And Batch


### Streaming

means that the Models are generating the content one by one and printing it too 
Instead of printing it together at a particular instance which is not good for user experience 

```python

for chunk in model.stream("Write a short paragraph about UPSC"):
    print(chunk.text, end=" ", flush=True)
```

```python
model.invoke("write a short paragraph about UPSC")
```

### Batch

batching a collection of independent requests to a model can significantly improve performance and reduce the cost by working parallel and generating the output all at once 

```python
responses = model.batch([
    "Why do we need AI Engineers",
    "Difference between AI Engineers and Web Dev",
    "who is Dep ops engineer"
],
    config={
        "max_concurrency": 2 # Limiting the max parallel can we can do 
    }
)
for response in responses:
    print(response)
    
```
