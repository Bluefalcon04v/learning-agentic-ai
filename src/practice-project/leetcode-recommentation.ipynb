# Leetcode Recommendation 

```python
import requests

url = "https://leetcode.com/graphql"

query = """
query recentSubmissions($username: String!) {
    recentAcSubmissionList(username: $username, limit: 5) {
        id
        title
        titleSlug
        timestamp
    }
}
"""

response = requests.post(
    url,
    json={
        "query": query,
        "variables": {
            "username": "amitsah04v"
        }
    },
    headers={
        "Content-Type": "application/json",
        "User-Agent": "Mozilla/5.0"
    }
)

data = response.json()

print(data)
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("ollama:llama3.2")

prompt = f"""
You are a helpful assistant for my DSA practice.

Based on my previously solved questions:
{data}

Recommend exactly 2 new LeetCode questions that I should solve next.

Consider:
- My previous problem-solving experience.
- Relevant DSA patterns.
- Gradually increasing difficulty.

Don't recommend questions I have already solved.

For each question, provide:
1. Problem name
2. Difficulty
3. DSA pattern
4. Why you recommend it
"""

response = llm.invoke(prompt)
print(response.content)
```

```python
# not needed for this script just to get the response in structured way
from pydantic import BaseModel, Field
from langchain.agents import create_agent

class ProfileInfo(BaseModel):
    """Get the data from the html parser"""
    
    name:str=Field(description="The name of the profile ")
    rank:str=Field(description="Get the Rank of the profile")
    submission_number:int=Field(description="Get the total number of submissions on the profile")
    solved_questions:list[str]=Field(description="Get the top 10 questions that have been solved on the profile")

agent = create_agent(
    model="gpt-4o-mini",
    response_format=ProfileInfo
)

response=agent.invoke({
    "messages":[
        {
            "role": "user",
            "content": f"Extract profile information from this data: {data}"
        }
    ]
})

structured_response = response["structured_response"]

print(structured_response)
```
