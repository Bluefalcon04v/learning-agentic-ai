# Messages

Messages are the fundamentals unit of context for models in LangChain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are object that contains.

- Role - Identifies the message type (eg: system or user)
- Content - Represents the actual content of the message (like text, images, audio, documents, etc)
- Metadata - Optional fields such as response information, message IDs, and token usage

LangChain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("ollama:llama3.2")
```

## <u>Text Prompt</u>
text prompts are strings - ideal for straightforward generation tasks where you don't need to retain conversation history.

```python
model.invoke("please tell me what is AI")

# use text prompts when:
#   - You have a single, standalone request
#   - You don't need conversation history
#   - You want minimal code complexity
```

## <u>Message Prompt</u>

Alternatively, you can pass in a list of messages to the model by providing a list of message objects.

<u>Message Type</u>
- System Message - Tells the model how to behave and provide context for interactions.
- Human Message - Represents user input and interactions with the model
- AI message -  Response generated by the model, including text content, tool calls and metadata
- Tool message - Represents the outputs of tool calls

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages=[
    SystemMessage("You are a poetry expert"),
    AIMessage("Hii im alexa AI Assistance"),
    HumanMessage("Write a poem on AI under 50 words")    
]

response = model.invoke(messages)
print(response)

print("-" *30)

responses1=model.stream(messages)
for response in responses1:
    print(response.text, end=" ", flush=False)
```

```python
# Detail information to the LLM through SystemMessage

from langchain.messages import SystemMessage, HumanMessage

system_message = SystemMessage("""
                               You are a senior Python developer with experience in AI 
                               Always provide the code block for your example 
                               be concise but through in your explanation.
                               and easy to read concept under 50 words
                               """)

messages = [
    system_message,
    HumanMessage("How do the AI works")
]

response=model.invoke(messages)
print(response.content)
```

```python
# Message Metadata

human_msg = HumanMessage(
    content="Hello world",
    name="Sam",
    id="msg_123"
)

response = model.invoke([human_msg])
response.content
```

```python
from langchain.messages import SystemMessage,AIMessage,HumanMessage

messages=[
    SystemMessage("You are a helpful AI assistance"),
    HumanMessage("Can you help me?"),
    AIMessage("Let me help you with this question"),
    HumanMessage("What is 14 + 21")    
]

response = model.invoke(messages)
response.content
```

## <u>Tool Message</u>


```python
from langchain.messages import AIMessage, ToolMessage

ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "Mumbai"},
        "id": "call_123"
    }]
)

weather_result = "Sunny"
tool_message= ToolMessage(
    content=weather_result,
    tool_call_id = "call_1324"
)

messages=[
        HumanMessage("what is weather in Mumbai"),
        ai_message,
        tool_message
]

response = model.invoke(messages)
response.content
```
