# Prompt Template 
we uses the prompt template because we don't want to send the complete prompts every time, that's why we use the prompt_template 

```python
# Example 1
from langchain_core.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables=["country"],
    template="What is the capital of the {country}"
)

res = prompt_template_name.format(country= "Pakistan")
print(res)
```

```python
# Example 2

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("What is the name of the company that makes this {products}")

prompt.format(products="Mac Book") # Printing the prompt
```

# Now how to use the PromptTemplate by Using Chains


```python
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

llm = init_chat_model(model="groq:qwen/qwen3.8-27b")

chains = prompt | llm | StrOutputParser()
print(chains.invoke("Mac Book"))
```

# Combining multiple chains 

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.chat_models import init_chat_model

llm = init_chat_model(model="groq:openai/gpt-oss-20b", temperature=0.4)

capital_prompt = PromptTemplate.from_template("Please tell me the capital of this {country}")
famous_food_template = PromptTemplate.from_template("give me 3 famous non-veg food of this {capital}")

capital_chain = capital_prompt | llm | StrOutputParser()
complete_chain = {"capital": capital_chain} | famous_food_template | llm | StrOutputParser()

res = complete_chain.invoke({"country": "india"})
print(res)
```
