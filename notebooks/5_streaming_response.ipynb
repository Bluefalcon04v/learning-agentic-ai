# Streaming

in the above the response get printed when it gets generated completely and then printed directly 
Now it will be generating one by one 

```python
llm = init_chat_model(model="groq:qwen/qwen3.8-27b", streaming=True)
```

```python
question = "What is AI?"
```

```python
res = llm.stream(question)
for chunk in res:
    print(chunk.content, end="")
```
