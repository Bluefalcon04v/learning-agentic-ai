```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("ollama:llama3.2")
```

# The Problem 

```python
text = "Hello my name is Bruce Banner, email id is abc@gmail.com and my number is 89039234"
```

```python
res = llm.invoke(f"Give me the name, email and number from this text: {text}")
response =  res.content
print(response.name) # this will not work because it is not in formatted structure
```

# Solution: Structured Output

```python
from pydantic import BaseModel, Field

class ResponseStructured(BaseModel):
    name: str = Field(description="Name of the User")
    email: str = Field(description="Email of the User")
    number: int = Field(description="Number of the User")
    
    
structured_llm = llm.with_structured_output(ResponseStructured)

```

```python
res = structured_llm.invoke(f"Hello my name is Bruce Banner, email id is abc@gmail.com and my number is 89039234")
print(type(res))
print(res)
print(res.model_dump())
```

```python
from pydantic import BaseModel, Field
from langchain.chat_models import init_chat_model

llm = init_chat_model("gpt-4o-mini")

class StructuredResponse(BaseModel):
    title: str = Field(description="Movie Title")
    release_year: int = Field(description="Movie Release Year")
    director: str = Field(description="Director name ")
    cast: list[str] = Field(description="List of the cast of the Movie")
    
structured_llm = llm.with_structured_output(StructuredResponse)
```

```python
res = structured_llm.invoke(f"Give me the one newly release movie of bollywood")
print(res.model_dump())
```
