# 1 Static Prompts

```python
from langchain.chat_models import init_chat_model
llm = init_chat_model(model="ollama:llama3.2")
```

```python
prompts= [
    {"role": "system", "content": "You are a translator and translate input in Hindi  "},
    {"role": "user", "content": "what is my name "},
]
```

```python
res = llm.invoke(prompts)
res.content
```

# 2 Prompt Template - Dynamic Prompts 
with the help of prompt template now we can dynamically use whatever we want to do rather than writing multiple line of code 

```python
from langchain_core.prompts import ChatPromptTemplate

prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator input in {language}"},
    {"role": "user", "content": "{query}"},
])
```

```python
final_prompts = prompts.format_messages(
    language= "french",
    query = " what is my name"
)

res = llm.invoke(final_prompts)
res.content
```

```python
res = prompts.invoke({
    "language": "german",
    "query": "Hello how are you?"
})

final_res = llm.invoke(res)
final_res


# instead of making this prompt -> llm -> final response 
# we can use the chains of langchain
```

# Chains 


```python
from langchain_core.output_parsers import StrOutputParser 
out = StrOutputParser() # this is just to get the content as a response from the llm 

def transform_case(res):
    return res.upper()
```

```python
chains = prompts | llm | out | transform_case
res = chains.invoke({
    "language": "german",
    "query": "Hello how are you?"
})
res
```
