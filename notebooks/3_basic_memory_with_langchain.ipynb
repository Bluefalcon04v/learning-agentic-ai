### Adding memory to the LLM 

```python
prompts = [
    {"role": "user", "content": "Hello my name is bruce"},
    {"role": "ai", "content": "Hello Bruce! It's nice to meet you. Is there something I can help you with, or would you like to chat for a bit ?"},
    {"role": "user", "content": "what did you said?"}
]

res = llm.invoke(prompts)

res.content
```

# Manage History

```python
history = []
while True:
    query = input("User: ")
    
    if query.lower() in ["exit", "quit", "bye"]:
        print("Goodbye ")    
        break
    
    history.append({"role": "user", "content": query})
    print("User: ", query, "\n")
    
    res = llm.invoke(history)
    history.append({"role": "ai", "content": res.content})
    print("AI: ", res.content, "\n")
    
```
